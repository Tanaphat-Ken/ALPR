# Resize real plate_lower images to 128x32

- Resizes all images from `../data/plate_lower/data` to 128x32.
- Upscaling uses `INTER_CUBIC`; downscaling falls back to `INTER_AREA`.
- Outputs to `../data/plate_lower/data_resized_128x32`.


```python
from pathlib import Path
import cv2

INPUT_DIR = Path("../data/plate_lower/data")
OUTPUT_DIR = Path("../data/plate_lower/data_resized_128x32")
TARGET_SIZE = (128, 32)  # (width, height)

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
IMAGE_EXTS = {".png", ".jpg", ".jpeg", ".bmp"}

def choose_interpolation(img, target_size=TARGET_SIZE):
    h, w = img.shape[:2]
    tw, th = target_size
    if w > tw or h > th:
        return cv2.INTER_AREA
    return cv2.INTER_CUBIC

def preprocess_real_data(input_dir=INPUT_DIR, output_dir=OUTPUT_DIR, target_size=TARGET_SIZE):
    if not input_dir.exists():
        raise FileNotFoundError(f"Input folder not found: {input_dir}")

    processed = 0
    for img_path in input_dir.iterdir():
        if img_path.suffix.lower() not in IMAGE_EXTS:
            continue

        img = cv2.imread(str(img_path))
        if img is None:
            print(f"⚠️ Skip unreadable file: {img_path.name}")
            continue

        interp = choose_interpolation(img, target_size)
        resized = cv2.resize(img, target_size, interpolation=interp)

        out_path = output_dir / img_path.name
        cv2.imwrite(str(out_path), resized)
        processed += 1

    print(f"✅ Processed {processed} images to {output_dir} at {target_size[0]}x{target_size[1]}.")
    print("Upscaling uses INTER_CUBIC; downscaling uses INTER_AREA.")

# Run the resize
preprocess_real_data()
```
